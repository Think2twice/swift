## 控制流程

```python
// 使用for-in循环数组
var mylist = [1,3,5,8,13]
for val in mylist {
    print(val)
}
// 循环哈希表
var playerScoreList = [
    "Curry": 31,
    "LeBron": 27,
    "Harden": 45
]
for score in playerScoreList {
    // 加入if-else判断
    if score.value >= 30 {
        print("\(score.key) is MVP.")
    } else {
        print("\(score.key) is good player.")
    }
}
```

```python
// 设置可以为空(nil)的变量
var age: Int? = 25
age = nil
if age == nil {
    print("年龄未知")
} else {
    print("年龄是：\(age)")
}
```

```python
// ??运算符
var fireball:String? = "火球"
var arrows = "万箭齐发"
print("对付骷髅海，首选的卡牌是：\(fireball ?? arrows)")

fireball = nil
print("对付骷髅海，首选的卡牌是：\(fireball ?? arrows)")
```

```python
// switch语法
var iris_p = 20
switch(iris_p) {
    case 0:
        print("山鸢尾")
    case 1:
        print("杂色鸢尾")
    case 2:
        print("维吉尼亚鸢尾")
    default:
        print("我没见过这花")
}
```

```python
// 各种循环的写法
// while写法
var i = 1
while i <= 5 {
    print("i:", i)
    i+=1
}
// repeat写法
var j = 10
repeat {
    print("j:", j)
    j+=1
} while j <= 15
// 自定义循环范围1
for k in 0..<3 {
    print("k:", k)
}
// 自定义循环范围2
for m in 0...3 {
    print("m:", m)
}
```

## 函数和闭包

```python
// 定义一个函数，参数，包括返回值
func sayHelo(name:String) -> String {
    return "helo \(name)."
}
print(sayHelo(name: "swift"))
print(sayHelo(name: "kotlin"))
```

```python
// 定义一个返回元组的函数
func calMath(x:Int, y:Int) -> (plus:Int, minus:Int, multi:Int, divide:Int) {
    return (x+y, x-y, x*y, x/y)
}
print(calMath(x: 20, y: 10))
// 获取计算结果，逐个输出
let result = calMath(x: 20, y: 10)
print(result.plus)
print(result.minus)
print(result.multi)
print(result.divide)
```

```python
// 嵌套定义函数
func calMath2(x:Int, y:Int) -> Int {
    func add() -> Int {
        return x + y
    }
    return add()
}
print(calMath2(x: 3, y: 4))
```

```python
// 返回一个函数
func calMath3() -> ((Int, Int)->Int) {
    func add(x:Int, y:Int) -> Int {
        return x+y
    }
    return add
}
var func_add = calMath3()
print(func_add(10, 20))
```

```python
// 定义回调函数
func showScreen(threshold:Int, cb: (String)->Void) {
    print("showScreen:", threshold)
    cb("helo u.")
}

func screen_cb(screen_data:String)->Void {
    print("screen_data is ", screen_data)
}
showScreen(threshold: 100, cb: screen_cb)
```

```python
// 闭包的使用
var numbers = [1, 2, 3, 4, 5]
var mylist = numbers.map({(number:Int) -> Int in
    return number * number
})
print(mylist)
// 正规的写法
func Power2(number:Int) -> Int{
    return number * number
}
mylist = numbers.map(Power2)
print(mylist)
// 简单的写法
mylist = numbers.map({number in number * number})
print(mylist)
// 排序闭包
mylist = numbers.sorted{$0>$1}
print(mylist)
```

## 对象和类

```python
// 定义一个类
class Player {
    var score = 0
    init(score: Int){
        self.score = score
    }
    func playerLevel() -> String {
        if self.score >= 30 {
            return "MVP"
        } else if self.score >= 20 {
            return "球星"
        } else if self.score >= 10 {
            return "首发"
        } else {
            return "板凳"
        }
    }
}

var player = Player(score: 25)
print(player.playerLevel())
```

```python
// 类继承，方法重载
class NBAPlayer: Player {
    override func playerLevel() -> String {
        if self.score >= 15 {
            return "交易过来"
        } else {
            return "交易出去"
        }
    }
}
var player = NBAPlayer(score: 23)
print(player.playerLevel())
```

```python
// 定义getter和setter
class Circle {
    // 半径
    var radius:Float = 0
    init(radius: Float) {
        self.radius = radius
    }
    // 面积
    var area: Float {
        get {
            return 3.14 * self.radius
        }
        set {
            self.radius = newValue / 3.14
        }
    }
}

var circle = Circle(radius: 10)
print(circle.area)

// 传入面积后，自动计算半径长度
circle.area = 62.8
print(circle.radius)
```

## 枚举和结构

```python
// 创建一个枚举类型
enum Lang {
    case java
    case python
    case javascript
    case typescript
    case go
    case vb
    case ruby
    case php
}

let mylang = Lang.java
switch(mylang) {
    case Lang.python:
        print("机器学习必备")
    case Lang.java:
        print("老牌经典语言")
    case Lang.go:
        print("并发你是第一")
    default:
        print("很不错的语言")
}
```

```python
// 定义枚举的自描述方法
enum Lang {
    case java
    case python
    case go
    case ruby
    func desc() {
        switch self {
            case .python:
                print("机器学习必备")
            case .java:
                print("老牌经典语言")
            case .go:
                print("并发你是第一")
            default:
                print("很不错的语言")
        }
    }
}
let mylang = Lang.python
mylang.desc()
```

```python
// 枚举的参数
enum HttpResponse {
    // 成功返回成功代码
    case success(Int)
    // 失败返回错误信息
    case failure(String)
}

// let result = HttpResponse.success(200)
let result = HttpResponse.failure("数据库访问失败")
switch result {
    case let .success(code):
        print("请求成功：", code)
    case let .failure(message):
        print("请求失败：", message)
}
```

## 协议和扩展

```python
// 定义一个协议(多数据库产品模式)
protocol IDatabase {
    var dbname:String { get set }
    func exeSql(sql:String) -> Int
    mutating func upgrade(dbname:String)
}

// PostgreSql
class PostgreSql: IDatabase {
    var dbname:String = "postgre"
    func exeSql(sql:String) -> Int {
        print("Postgre执行Sql：", sql)
        return 100
    }
    func upgrade(dbname:String) {
        self.dbname = dbname
    }

}
// MySql
class MySql: IDatabase {
    var dbname:String = "mysql"
    func exeSql(sql:String) -> Int {
        print("MySql执行Sql：", sql)
        return 200
    }
    func upgrade(dbname:String) {
        self.dbname = dbname
    }
}

let db1 = PostgreSql()
var result1 = db1.exeSql(sql: "SELECT * FROM TABLES")
db1.upgrade(dbname: "PostgreSql 12.4")
print(db1.dbname)

let db2 = MySql()
var result2 = db2.exeSql(sql: "DESC TABLES")
db2.upgrade(dbname: "MySql 8.0.21")
print(db2.dbname)
```

```python
// 扩展PostgreSql类，添加新功能
extension PostgreSql {
    func batchExeSql(sql:[String]) -> Int {
        print("批处理Sql", sql)
        return 300
    }
}
let db3 = PostgreSql()
db3.batchExeSql(sql: ["BEGIN TRANSACTION ... ", "UPDATE TABLE1 SET ..."])
```

## 错误处理

```python
// 定义错误枚举
enum DatabaseError: Error {
    case db_error
    case sql_error
    case network_error
}
// 定义一个抛例外的方法
func exeSqlCom(sql:String?) throws -> Int {
    if sql == nil {
        throw DatabaseError.sql_error
    }
    return 1
}
// 例外捕捉
do {
    try exeSqlCom(sql: nil)
} catch DatabaseError.sql_error {
    print("sql出错")
} catch {
    print(error)
}
// try?例外捕捉
let db_result = try? exeSqlCom(sql: nil)
print(db_result)
```
